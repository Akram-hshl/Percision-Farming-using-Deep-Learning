```python
import os
import random
import csv
from shutil import copyfile

#image_folder = r'C:\Users\Richa\Downloads\Competition_Test\NonsegmentedV2'
#image_folder = '/raid10/active_users/groupC/ml/dautogeneratedBinFolder/pwfg/f/dataset/NonsegmentedV2'

image_folder = r'C://Users//$8CUJ00-BGKRQC8B3RAL//Desktop//Plant Lab//NonsegmentedV2'

sample_size = 75

# here are stored the images as it was specified by Mr. Rother
#sampled_dataset_path = 'C:/Users/Richa/Downloads/Competition_Test/comp_prep'

sampled_dataset_path ='C://Users//$8CUJ00-BGKRQC8B3RAL//Desktop//Plant Lab//Competetion data7//Competation_image7'
os.makedirs(sampled_dataset_path, exist_ok=True)

csv_file = open('competition_groupA7.csv', 'w', newline='')
csv_writer = csv.writer(csv_file, delimiter=';') 
csv_writer.writerow(['id', 'label'])




image_counter = 0

for subfolder in os.listdir(image_folder):
    subfolder_path = os.path.join(image_folder, subfolder)
    
    if os.path.isdir(subfolder_path):
        images = os.listdir(subfolder_path)
  
        sampled_images = random.sample(images, sample_size)

        for image in sampled_images:
            src_path = os.path.join(subfolder_path, image)
            dst_path = os.path.join(sampled_dataset_path, f'{image_counter}.png')
            copyfile(src_path, dst_path)
            
            csv_writer.writerow([image_counter, subfolder])
            
            image_counter += 1

csv_file.close()
```

```python
import zipfile

zip_filename = 'comp_prep7.zip'
#when running it in the server:
##'/raid10/active_users/groupC/ml/competition_groupC/comp_prep.zip'

with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for root, dirs, files in os.walk(sampled_dataset_path):
        for file in files:
            file_path = os.path.join(root, file)
            arcname = os.path.relpath(file_path, sampled_dataset_path)
            zipf.write(file_path, arcname)
```
